Household Coverage Percentage Analysis

The household coverage percentage column is examined to understand the distribution of reported household coverage under the Swachh Bharat Mission (Grameen) solid waste management initiative. The describe() function provides statistical summaries, including the mean, median, minimum, maximum, and standard deviation. The missing-value check identifies records where coverage information is unavailable. This analysis helps assess the completeness of the data and understand variations in reported household coverage across the dataset.

```python
coverage_col = "Households Covered(%) (hhs_covered_per)"

print(df[coverage_col].describe())
print("Missing coverage values:", df[coverage_col].isna().sum())
```

Description: Identifying Missing Coverage Values

This code creates a copy of the original dataset and adds a new column named coverage_missing to identify rows where household coverage percentage values are missing. The isna() function returns True for missing values and False for available values. Finally, value_counts() counts how many rows have missing and non-missing coverage values.

```python
df_clean = df.copy()

# Keep track of rows with missing coverage
df_clean["coverage_missing"] = (
    df_clean[coverage_col].isna()
)

print(df_clean["coverage_missing"].value_counts())
```

Description: Removing Duplicate Rows

This code identifies the number of exact duplicate rows in the dataset using duplicated().sum(). The drop_duplicates() function removes duplicate rows while keeping the first occurrence of each unique row. Finally, shape displays the number of rows and columns remaining after removing duplicates, helping ensure the dataset is free from repeated records.

```python
# Count exact duplicates
print("Duplicate rows:", df_clean.duplicated().sum())

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Shape after removing duplicates:", df_clean.shape)
```

 Checking Repeated Village Codes

This code identifies the number of repeated village codes in the dataset using the duplicated().sum() function. The village_code column is selected to check whether any village codes appear more than once. This helps determine whether the dataset contains multiple records associated with the same village. Repeated codes do not necessarily indicate duplicate rows, as a village may have multiple records.

```python
village_code = "Village Code (village_code)"

print(
    "Repeated village codes:",
    df_clean[village_code].duplicated().sum()
)
```

```python
print(df_clean.dtypes)
```

 Correcting Data Types and Cleaning Text Columns

This code cleans the text columns by converting their values to Pandas string data type and removing unnecessary leading and trailing spaces using str.strip(). It also converts the household coverage column into a numeric data type using pd.to_numeric(). The errors="coerce" parameter converts invalid or non-numeric values into missing values (NaN). Finally, df_clean.dtypes displays the data type of each column to verify that the conversions were applied correctly.

```python
text_cols = [
    "State Name (state_name)",
    "District Name (district_name)",
    "Block Name (block_name)",
    "Gram Panchayat Name (gp_name)",
    "Village Name (village_name)",
    "Asset (asset)"
]

for col in text_cols:
    df_clean[col] = df_clean[col].astype("string").str.strip()

# Ensure coverage is numeric
df_clean[coverage_col] = pd.to_numeric(
    df_clean[coverage_col], errors="coerce"
)

print(df_clean.dtypes)
```

Identifying Invalid Coverage Values

This code identifies records where the household coverage percentage is less than 0 or greater than 100. The conditions are combined using the | operator, which represents logical OR. The matching records are stored in invalid_coverage, and len() counts the number of invalid records. Finally, display() shows the identified records for further inspection and data cleaning.

```python
invalid_coverage = df_clean[
    (df_clean[coverage_col] < 0) |
    (df_clean[coverage_col] > 100)
]

print("Invalid coverage records:", len(invalid_coverage))
display(invalid_coverage)
```

Statistical Summary of Numeric Columns

This code selects all numeric columns from the cleaned dataset using select_dtypes(include=np.number). The describe() function calculates descriptive statistics, including count, mean, standard deviation, minimum, maximum, and quartile values for each numeric column. The .T attribute transposes the output so that each numeric column appears as a row, making it easier to compare and understand the statistical characteristics of the data. Finally, display() presents the summary in a readable tabular format.

```python
numeric_cols = df_clean.select_dtypes(
    include=np.number
).columns

display(df_clean[numeric_cols].describe().T)
```

Creating Coverage Level Categories

This code creates a new column named Coverage Level to classify household coverage percentages into four categories: Low (0–25%), Moderate (25–50%), High (50–75%), and Very High (75–100%). The pd.cut() function divides the numeric coverage values into the specified ranges and assigns a corresponding label to each value. Finally, display() shows the first 10 rows containing the original coverage percentages and their assigned categories, making it easier to understand and compare household coverage levels.

```python
df_clean["Coverage Level"] = pd.cut(
    df_clean[coverage_col],
    bins=[-0.01, 25, 50, 75, 100],
    labels=[
        "Low (0-25%)",
        "Moderate (25-50%)",
        "High (50-75%)",
        "Very High (75-100%)"
    ]
)

display(
    df_clean[
        [coverage_col, "Coverage Level"]
    ].head(10)
)
```

Identifying Recorded Coverage Values

This code creates a new column named Coverage Recorded to indicate whether household coverage data is available for each record. The np.where() function assigns "Yes" when the coverage value is present and "No" when it is missing. Finally, value_counts() counts the number of records with and without coverage information, helping assess the completeness of the dataset.

```python
df_clean["Coverage Recorded"] = np.where(
    df_clean[coverage_col].notna(),
    "Yes",
    "No"
)

print(df_clean["Coverage Recorded"].value_counts())
```

Filtering High Coverage Records

This code filters the dataset to identify records where household coverage is greater than or equal to 75%. The condition df_clean[coverage_col] >= 75 selects records that meet this criterion, and the results are stored in the high_coverage DataFrame. The len() function counts the number of records with high coverage, while display(high_coverage.head()) displays the first five matching records for further analysis.

```python
high_coverage = df_clean[
    df_clean[coverage_col] >= 75
]

print("Records with coverage >= 75%:", len(high_coverage))
display(high_coverage.head())
```

State-Wise Household Coverage Analysis

This code groups the dataset by state name and calculates summary statistics for household coverage percentages in each state. The groupby() function groups records by state, while agg() calculates the count, mean, median, minimum, and maximum coverage values for each state. The round(2) function rounds the results to two decimal places, and sort_values("mean", ascending=False) sorts the states in descending order based on their average household coverage. Finally, display() presents the state-wise summary in a table, helping identify states with higher and lower average coverage levels.

```python
state_col = "State Name (state_name)"

state_summary = (
    df_clean.groupby(state_col)[coverage_col]
    .agg(["count", "mean", "median", "min", "max"])
    .round(2)
    .sort_values("mean", ascending=False)
)

display(state_summary)
```

Exporting the Cleaned Dataset

This code saves the cleaned dataset as a CSV file named Solid_Waste_Management_Cleaned.csv. The to_csv() function exports the DataFrame to a file, while index=False prevents Pandas from adding the row index as an extra column. The success message confirms that the export command has completed, and df_clean.shape displays the final number of rows and columns in the cleaned dataset. This step makes the processed data available for future analysis, visualization, and reporting.

```python
df_clean.to_csv(
    "Solid_Waste_Management_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("Final shape:", df_clean.shape)
```
